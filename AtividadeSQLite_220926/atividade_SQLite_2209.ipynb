# Atividade SQLite

## Parte 1:

```python
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("empresa.db")
cursor = conn.cursor()
```

Código comentado para restaurar os dados:

### Criação de tabela (DDL):

```python
cursor.execute("""
CREATE TABLE Funcionarios (
ID INTEGER PRIMARY KEY AUTOINCREMENT,
Nome TEXT NOT NULL,
Cargo TEXT NOT NULL,
Salario REAL,
Data_Contratacao TEXT NOT NULL
);
""")
```

Verificação da criação da tabela:

```python
cursor.execute("SELECT name FROM sqlite_master;")
print(cursor.fetchall())
```

## Parte 2:

### Manipulação de Dados (DML):

```python
cursor.executemany("INSERT INTO Funcionarios (Nome,Cargo, Salario, Data_Contratacao) VALUES (?,?,?,?)", [
('Lucas Almeida', 'Analista de Dados',4200,'2024-03-15'),
('Mariana Souza', 'Desenvolvedora Backend',5100,'2023-08-21'),
('Rafael Mendes', 'Técnico de Redes',3500,'2025-01-10'),
('Camila Ferreira', 'Analista de Suporte',3200,'2024-11-05'),
('Bruno Carvalho', 'Coordenador de TI',6800,'2022-06-18')
])
```

### Consulta Geral do Banco:

```python
cursor.execute(""" SELECT * FROM Funcionarios """)
pd.DataFrame(cursor.fetchall())
```

### Consulta Filtrada do Banco:
Condições do Filtro:
- Selecione apenas os funcionários com salário superior a R$ 5.000.
- Selecione os funcionários que trabalham como 'Gerente'.

```python
cursor.execute(""" SELECT * FROM Funcionarios WHERE Salario > 5000""")
pd.DataFrame(cursor.fetchall())
```

```python
cursor.execute(""" SELECT * FROM Funcionarios WHERE Cargo == 'Gerente' """)
pd.DataFrame(cursor.fetchall())
```

### Atualização de dados (UPDATE):

```python
cursor.execute(""" UPDATE Funcionarios  SET Salario = Salario * 1.10 WHERE Nome == 'Lucas Almeida' """)
pd.DataFrame(cursor.fetchall())
```

Verificação do UPDATE:

```python
cursor.execute(""" SELECT * FROM Funcionarios WHERE Nome == 'Lucas Almeida' """)
pd.DataFrame(cursor.fetchall())
```

### Exclusão de dados (DELETE):

```python
cursor.execute(""" DELETE FROM Funcionarios WHERE Nome = 'Bruno Carvalho'; """)
```

Verificação de DELETE:

```python
cursor.execute(""" SELECT * FROM Funcionarios """)
pd.DataFrame(cursor.fetchall())
```

### Contagem de registros no banco:

```python
cursor.execute(""" SELECT COUNT(ID) FROM Funcionarios """)
cursor.fetchone()
```

## Parte 3: Agregação e Ordenação

### Funções de agregação ( SUM, AVG, MAX, MIN):

Verificação da soma dos salários:

```python
cursor.execute(""" SELECT SUM(Salario) FROM Funcionarios """)
cursor.fetchone()
```

Verificação da média dos salários:

```python
cursor.execute(""" SELECT AVG(Salario) FROM Funcionarios """)
cursor.fetchone()
```

Verificação do maior salário:

```python
cursor.execute(""" SELECT MAX(Salario) FROM Funcionarios """)
cursor.fetchone()
```

Verificação do menor salário:

```python
cursor.execute(""" SELECT MIN(Salario) FROM Funcionarios """)
cursor.fetchone()
```

###  Agrupamento de dados ( GROUP BY):

Selecionando todos os funcionários e ordens de maior para o menor salário:

```python
cursor.execute(""" SELECT Cargo, AVG (Salario) FROM Funcionarios GROUP BY Cargo""")
cursor.fetchall()
```

Selecione todos os funcionários e ordene-os por nome em ordem alfabética:

```python
cursor.execute(""" SELECT Cargo, AVG (Salario) FROM Funcionarios GROUP BY Cargo""")
cursor.fetchall()
```

## Parte 4: Transações e ROLLBACK:

### Simulação de transação:

Tentativa com simulação de erro:

```python
try:
    conn.execute("BEGIN;")

    cursor.execute(""" INSERT INTO Funcionarios (Nome, Cargo, Salario, Data_Contratacao) VALUES ('Gerson Santana','Analista de Redes',7500 ,'2026-09-22') """)

    raise Exception("Erro simulado!")

    conn.commit()
except:
    conn.rollback()
    print("Transação revertida!")
```

Transação normal:

```python
try:
    conn.execute("BEGIN;")

    cursor.execute(""" INSERT INTO Funcionarios (Nome, Cargo, Salario, Data_Contratacao) VALUES ('Gerson Santana','Analista de Redes',7500 ,'2026-09-22') """)

    conn.commit()
except:
    conn.rollback()
    print("Transação revertida!")
```

```python
cursor.execute(" SELECT * FROM Funcionarios ")
pd.DataFrame(cursor.fetchall())
```
